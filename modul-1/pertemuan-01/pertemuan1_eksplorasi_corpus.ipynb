```python
# ==========================================
# 1. IMPORT LIBRARY
# ==========================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import nltk
from nltk.tokenize import sent_tokenize
from IPython.display import display

# Mengunduh resource NLTK untuk tokenisasi kalimat
nltk.download("punkt")
nltk.download("punkt_tab")


```

```python
# ==========================================
# 2. MEMBACA DATASET
# ==========================================
df = pd.read_csv('MBG_Arya.csv')

print("\n--- Informasi Dataset ---")
print(df.info())
```

```python
# ==========================================
# 3. INSPEKSI DAN EKSPLORASI AWAL DATASET
# ==========================================
# Menampilkan 10 data teratas dan terbawah
print("--- 10 Data Teratas ---")
display(df.head(10))

print("\n--- 10 Data Terbawah ---")
display(df.tail(10))

# Informasi struktur dan ukuran dataset
print("\nUkuran dataset (baris, kolom):", df.shape)
print("\nNama kolom:")
print(df.columns.tolist())

print("\nInformasi tipe data dan Non-Null count:")
df.info()

# Menghitung total seluruh dokumen/baris
jumlah_dokumen = len(df)
print("\nJumlah total dokumen:", jumlah_dokumen)
```

```python
# ==========================================
# 4. SAMPLING SAMPEL TEKS AWAL
# ==========================================
text_column = "text_original"
texts = df[text_column]

print("\n--- Sampel 10 Teks Pertama ---")
for text in texts.head(10):
    print(text)
    print("-" * 80)
```

```python
# ==========================================
# 5. PEMERIKSAAN & PENANGANAN MISSING VALUES
# ==========================================
print("\n--- Pemeriksaan Missing Values ---")
print("Missing values per kolom:")
print(df.isna().sum())
print("Total missing value di seluruh dataset:", df.isna().sum().sum())

# Menghapus baris yang memiliki nilai kosong pada kolom teks utama
df = df.dropna(subset=[text_column])

# Memastikan nilai kosong pada kolom teks sudah 0
print("Sisa missing value pada kolom teks:", df[text_column].isnull().sum())
```

```python
# ==========================================
# 6. PEMERIKSAAN DATA DUPLIKAT
# ==========================================
print("\n--- Pemeriksaan Duplikasi Data ---")
print("Jumlah baris duplikat utuh (seluruh kolom sama):", df.duplicated().sum())
print("Jumlah komentar duplikat (kolom teks sama):", df.duplicated(subset=text_column).sum())
```

```python
# ==========================================
# 7. EKSTRAKSI FITUR TEKS (PANJANG TEKS)
# ==========================================
# A. Menhitung jumlah karakter dalam teks
df["jumlah_karakter"] = df[text_column].astype(str).apply(len)

# B. Menhitung jumlah kata dalam teks
df["jumlah_kata"] = df[text_column].astype(str).apply(lambda x: len(x.split()))

# C. Menhitung jumlah kalimat dalam teks menggunakan NLTK
df["jumlah_kalimat"] = df[text_column].astype(str).apply(lambda x: len(sent_tokenize(x)))

# Menampilkan statistik deskriptif panjang kata
print("\n--- Statistik Deskriptif Jumlah Kata ---")
print(df["jumlah_kata"].describe())
```

```python
# ==========================================
# 8. RINGKASAN STATISTIK DESKRIPTIF TEKS
# ==========================================
texts_clean = df[text_column].fillna("").astype(str)

print("\n--- Ringkasan Statistik Teks ---")
print("Jumlah dokumen bersih:", len(df))
print("Jumlah komentar tidak kosong:", texts_clean.str.strip().ne("").sum())
print("Total seluruh karakter:", texts_clean.str.len().sum())
```

```python
# ==========================================
# 9. VISUALISASI DISTRIBUSI TEKS
# ==========================================
# A. Visualisasi Distribusi Jumlah Kata
plt.figure(figsize=(10, 5))
plt.hist(df["jumlah_kata"], bins=30, color="steelblue", edgecolor="white")
plt.xlabel("Jumlah Kata")
plt.ylabel("Jumlah Dokumen")
plt.title("Distribusi Panjang Dokumen (Jumlah Kata)")
plt.tight_layout()
plt.show()

# B. Frekuensi Jumlah Kalimat per Dokumen
print("\n--- Frekuensi Jumlah Kalimat ---")
print(df["jumlah_kalimat"].value_counts())

# C. Visualisasi Distribusi Jumlah Kalimat
plt.figure(figsize=(10, 5))
df["jumlah_kalimat"].value_counts().plot(kind="bar", color="steelblue")
plt.xlabel("Jumlah Kalimat")
plt.ylabel("Jumlah Dokumen")
plt.title("Distribusi Jumlah Kalimat")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 10. VERIFIKASI AKHIR MISSING VALUES
# ==========================================
print("\n--- Verifikasi Akhir Missing Values ---")
print(df.isna().sum())
```
