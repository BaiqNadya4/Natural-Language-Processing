```python
import pandas as pd
import time

data = pd.read_csv("MBG_Arya.csv")
print("Kolom dataset:", data.columns.tolist())

# Dataset sudah memiliki hasil cleaning dan preprocessing pada kolom ini.
text_column = "processed_text"
batch_size = 200
processed_tokens = []

start_time = time.time()

for start in range(0, len(data), batch_size):
    end = min(start + batch_size, len(data))
    batch = data[text_column].iloc[start:end]
    hasil = batch.fillna("").astype(str).str.split()
    processed_tokens.extend(hasil.tolist())
    print(f"Data {end} dari {len(data)} selesai diproses")

data["processed_tokens"] = processed_tokens

print("\nPreprocessing selesai!")
print(f"Waktu proses: {time.time() - start_time:.2f} detik")
print(data[["text_original", "processed_tokens"]].head(10))
```

```python
# ==========================================
# 1. IMPORT LIBRARY
# ==========================================
import re
import string
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
# ==========================================
# 2. LOAD DATASET
# ==========================================
df = pd.read_csv('MBG_Arya.csv')

# Gunakan kolom 'text_original'
text_column = "text_original"

# Hapus missing values pada kolom text_original
df = df.dropna(subset=[text_column])
print(f"Jumlah data siap di-normalisasi: {len(df)}")
```

```python
# ==========================================
# 3. FUNGSI CLEANING TEKS
# ==========================================
def case_folding(text):
    return str(text).lower()

def remove_url(text):
    return re.sub(r"http\S+|www\S+", "", text)

def remove_mention(text):
    return re.sub(r"@\w+", "", text)

def remove_hashtag(text):
    return re.sub(r"#\w+", "", text)

def remove_number(text):
    return re.sub(r"\d+", "", text)

def remove_punctuation(text):
    return text.translate(str.maketrans("", "", string.punctuation))

def clean_whitespace(text):
    return re.sub(r"\s+", " ", text).strip()

def clean_text(text):
    text = case_folding(text)
    text = remove_url(text)
    text = remove_mention(text)
    text = remove_hashtag(text)
    text = remove_number(text)
    text = remove_punctuation(text)
    text = clean_whitespace(text)
    return text

# Contoh penggunaan fungsi clean_text
contoh_teks = "Halo @budi! Cek https://contoh.com #NLP2026"
hasil_cleaning = clean_text(contoh_teks)
print("Teks sebelum cleaning:", contoh_teks)
print("Teks setelah cleaning:", hasil_cleaning)
```

```python
# ==========================================
# 4. STOPWORD REMOVAL & STEMMING SETUP
# ==========================================
stop_words = set(stopwords.words("indonesian"))

# Inisialisasi Stemmer Sastrawi
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def remove_stopwords(tokens):
    return [token for token in tokens if token not in stop_words]

def stem_tokens(tokens):
    return [stemmer.stem(token) for token in tokens]

# Contoh penggunaan remove_stopwords dan stem_tokens
contoh_tokens = ["saya", "sedang", "mempelajari", "pemrosesan", "bahasa", "alami"]

tanpa_stopword = remove_stopwords(contoh_tokens)
hasil_stemming = stem_tokens(tanpa_stopword)

print("Token awal      :", contoh_tokens)
print("Tanpa stopword  :", tanpa_stopword)
print("Hasil stemming :", hasil_stemming)
```

```python
# ==========================================
# 5. PREPROCESSING PIPELINE LENGKAP
# ==========================================
def preprocessing_pipeline(text):
    # 1. Cleaning
    cleaned_text = clean_text(text)
    # 2. Tokenization
    tokens = word_tokenize(cleaned_text)
    # 3. Stopword Removal
    tokens_no_stopwords = remove_stopwords(tokens)
    # 4. Stemming
    stemmed_tokens = stem_tokens(tokens_no_stopwords)
    return stemmed_tokens

# Contoh penggunaan preprocessing_pipeline
contoh_teks = "Saya sedang mempelajari pemrosesan bahasa alami di Indonesia."
hasil_pipeline = preprocessing_pipeline(contoh_teks)

print("Teks awal:", contoh_teks)
print("Hasil pipeline:", hasil_pipeline)
print("Teks hasil:", " ".join(hasil_pipeline))
```

```python
# ==========================================
# 6. PENERAPAN KE DATASET
# ==========================================
print("\nMemproses dataset... (Membutuhkan beberapa menit untuk stemming)")

# Terapkan fungsi cleaning dasar
df["clean_text"] = df[text_column].apply(clean_text)

# Terapkan pipeline lengkap
df["processed_tokens"] = df[text_column].apply(preprocessing_pipeline)

# Gabungkan token kembali menjadi kalimat
df["processed_text"] = df["processed_tokens"].apply(lambda tokens: " ".join(tokens))

print("Preprocessing selesai!")
```

```python
# ==========================================
# 7. PERBANDINGAN SEBELUM VS SESUDAH
# ==========================================
print("\n=== PERBANDINGAN RAW VS PROCESSED TEXT ===")
for i in range(min(5, len(df))):
    print(f"RAW ({text_column}) : {df.iloc[i][text_column]}")
    print(f"CLEAN TEXT              : {df.iloc[i]['clean_text']}")
    print(f"PROCESSED TEXT          : {df.iloc[i]['processed_text']}")
    print("=" * 80)
```

```python
# ==========================================
# 8. MENYIMPAN DATASET HASIL PREPROCESSING
# ==========================================
output_columns = [text_column, "clean_text", "processed_text"]

import os
os.makedirs("../output", exist_ok=True)

output_path = "../output/dataset_preprocessed.csv"
df[output_columns].to_csv(output_path, index=False)

print(f"\nDataset preprocessed berhasil disimpan di: {output_path}")
```
